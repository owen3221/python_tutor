# Python Web Scraping Tutorial

Web scraping means writing a program to collect information from a webpage.

For example, instead of copying titles or prices by hand, we can use Python to download the webpage and extract the data we need.

Before scraping a website, always check:

1. Do not overload the website.
2. Do not collect private information.
3. Follow the website rules.
4. Use scraping only for learning or allowed data.

# Basic idea of web scraping:
1. Send a request to a webpage
2. Get the HTML
3. Extract useful information

## 2. Sending a Request to a Webpage

A webpage is usually written in HTML.

Python can download the HTML using `requests.get()`.

Important idea:

```python
response = requests.get(url)

````python
import requests

url = "https://example.com"
response = requests.get(url)

print(response.status_code)
print(response.text[:200])
````

## 3. Checking the Status Code

A status code tells us whether the request was successful.

Common status codes:

- `200`: success
- `404`: page not found
- `403`: forbidden
- `500`: server error

Usually, we only continue if the status code is `200`.

````python
import requests

url = "https://example.com"
response = requests.get(url)

if response.status_code == 200:
    print("Request successful.")
else:
    print("Request failed:", response.status_code)
````

## 4. Understanding HTML

HTML uses tags to structure a webpage.

Example:

```html
<h1>Main Title</h1>
<p>This is a paragraph.</p>
<a href="https://example.com">A link</a>
```

Common tags:

* `<h1>`: main heading
* `<p>`: paragraph
* `<a>`: link
* `<div>`: section
* `<span>`: small text area

````python


html = """
<html>
    <body>
        <h1>My Website</h1>
        <p>Hello students.</p>
    </body>
</html>
"""

print(html)
````

## 5. Parsing HTML with BeautifulSoup

BeautifulSoup helps Python understand HTML.

Basic syntax:

```python
soup = BeautifulSoup(html_text, "html.parser")

````python
from bs4 import BeautifulSoup

html = """
<html>
    <body>
        <h1>My Website</h1>
        <p>Hello students.</p>
    </body>
</html>
"""

soup = BeautifulSoup(html, "html.parser")

title = soup.find("h1")
paragraph = soup.find("p")

print(title.text)
print(paragraph.text)
````

## 6. Finding One Element with `find()`

`find()` finds the first matching tag.

Example:

```python
soup.find("h1")

````python

from bs4 import BeautifulSoup

html = """
<h1>First Title</h1>
<h1>Second Title</h1>
"""

soup = BeautifulSoup(html, "html.parser")

first_title = soup.find("h1")

print(first_title.text)
````

## 7. Finding Multiple Elements with `find_all()`

`find_all()` finds all matching tags.

Example:

```python
soup.find_all("p")

````python

from bs4 import BeautifulSoup

html = """
<p>Apple</p>
<p>Banana</p>
<p>Cherry</p>
"""

soup = BeautifulSoup(html, "html.parser")

items = soup.find_all("p")

for item in items:
    print(item.text)
````

## 8. Getting Text with `.text`

HTML tags contain text.

Example:

```html
<h1>Hello</h1>

````python
from bs4 import BeautifulSoup

html = "<h1>Python Web Scraping</h1>"

soup = BeautifulSoup(html, "html.parser")
title = soup.find("h1")

print(title.text)
````

## 9. Getting Links with `href`

Links usually use the `<a>` tag.

Example:

```html
<a href="https://example.com">Click here</a>

````python
from bs4 import BeautifulSoup

html = """
<a href="https://example.com">Example Website</a>
"""

soup = BeautifulSoup(html, "html.parser")

link = soup.find("a")

print(link.text)
print(link["href"])
````

## 10. Scraping a Simple Webpage

Now we combine everything:

1. Use `requests.get()`
2. Check status code
3. Use BeautifulSoup
4. Extract data

````python
import requests
from bs4 import BeautifulSoup

url = "https://example.com"
response = requests.get(url)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")

    title = soup.find("h1")
    paragraph = soup.find("p")

    print("Title:", title.text)
    print("Paragraph:", paragraph.text)
else:
    print("Failed to download page.")
````

## 11. Saving Scraped Data to a File

After collecting data, we may want to save it.

We can use:

- `open`
- `write`
- `with`
- `\n`

This is useful because scraped data is often stored for later analysis.

````python
import requests
from bs4 import BeautifulSoup

url = "https://example.com"
response = requests.get(url)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    title = soup.find("h1").text

    with open("scraped_title.txt", "w", encoding="utf-8") as f:
        f.write(title + "\n")

    print("Title saved.")
else:
    print("Failed to download page.")
````

## 12. Handling Errors with `try`

Sometimes scraping fails because:

- Internet connection is bad
- The URL is wrong
- The website blocks the request
- The page structure changes

We can use `try` and `except` to avoid crashing.

````python
import requests
from bs4 import BeautifulSoup

url = "https://example.com"

try:
    response = requests.get(url)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, "html.parser")
        title = soup.find("h1").text
        print(title)
    else:
        print("Request failed:", response.status_code)

except requests.exceptions.RequestException:
    print("Network error occurred.")
````

## 13. Adding Headers

Some websites may reject requests that do not look like they come from a browser.

We can add a simple header.

A header gives extra information about the request.

````python
import requests

url = "https://example.com"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

print(response.status_code)
print(response.text[:100])
````
